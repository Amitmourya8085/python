```python
###### import numpy as np

marks = np.array([
    [85, 90, 78],
    [35, 45, 50],
    [95, 88, 92],
    [60, 65, 70],
    [25, 30, 35],
    [75, 80, 85]
])
print(np.where(marks>=40,"pass","fail"))
conditions = [
    marks >= 75,
    (marks >= 60) & (marks < 75),
    (marks >= 40) & (marks < 60),
    marks < 40
]

choices = ["Excellent", "Good", "Average", "Fail"]

grades = np.select(conditions, choices)

print(grades)
```
