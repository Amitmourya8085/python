```python
import numpy as np

arr= np.array([1,2,3,4])
b= arr

b[0]=100
print("original:",arr)
print("new:",b)
# not create new array its refer same array
```

```python
arr = np.array([10, 20, 30, 40, 50])

b = arr.copy() # copy() create separate array

b[0] = 100

print("Original:", arr)
print("Copy:", b)
```

```python
marks = np.array([45, 78, 32, 90, 56, 88, 41])

marks[marks < 40] = 40#boolean masing

print(marks)

result=np.where(marks >= 50,0,1)
print(result)
```

```python
#Q2
marks = np.array([45, 78, 32, 90, 56, 88, 41, 67, 95, 29])
updated_marks=marks.copy()
updated_marks[updated_marks <40]=40
updated_marks[updated_marks >90]=90
print(f"original={marks};updated={updated_marks}")
print(marks.mean()+updated_marks.mean()/2)

```
