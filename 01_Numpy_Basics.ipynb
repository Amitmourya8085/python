```python
import numpy as np

arr = np.array([12, 5, 8, 20, 5, 15, 20, 3, 8, 10])


print("Maximum value:", np.max(arr))
print("Minimum value:", np.min(arr))
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard deviation:", np.std(arr))
print("Sum:", np.sum(arr))
print("Unique values:", np.unique(arr))
print("Index of maximum value:", np.argmax(arr))
print("Sort the array:", np.sort(arr))

data = np.arange(1, 25)


matrix = data.reshape(4, 6)

print("Shape:", matrix.shape)
print("Mean (complete):", np.mean(matrix))
print("Maximum:", np.max(matrix))
print("Minimum:", np.min(matrix))
print("Mean of each row:", np.mean(matrix, axis=1))
print("Mean of each column:", np.mean(matrix, axis=0))
print("Flattened:", matrix.flatten())
```
