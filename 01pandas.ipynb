```python
#selecting& accessing
import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Priya", "Rohan", "Neha"],
    "Age": [20, 21, 20, 22, 19],
    "Marks": [85, 76, 95, 68, 88],
    "City": ["Boisar", "Mumbai", "Palghar", "Vasai", "Thane"]
}

df = pd.DataFrame(data)

print(df)
print(df.columns)
print(df['Name'])
print(df[["Name","Age"]])
print("index 0=",df.loc[0])
print("index 0 and Name:",df.loc[0,"Name"])
print("index 0:2:",df.loc[0:2,["Name","Age"]])
print("idex 0,2",df.iloc[0,2])

#filtering 
print(df[df["Marks"]<80])
mul_fil=df[df["Marks"]>80 &(df['Age']==20)]
print(mul_fil)
```
