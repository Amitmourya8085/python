```python
import numpy as np

arr = np.array([4, 9, 16, 25, 36])
print(np.sqrt(arr))
print(np.square(arr))
print(np.power(arr,4))
print(np.round(np.exp(arr)))
print(np.round(np.log(arr)))

arrr = np.array([99.456, 149.789, 250.234, 399.999])
print(np.round(arrr))
print(np.floor(arrr))
print(np.ceil(arrr))

a = np.array([
    [1, 2],
    [3, 4]
])

print(np.dot(a, a))
print(np.transpose(a))
print(np.linalg.det(a))
print(np.linalg.inv(a))
```
